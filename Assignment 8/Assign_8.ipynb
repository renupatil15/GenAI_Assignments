## **Roman Minglish → Devanagari Marathi**

```python
from indic_transliteration import sanscript
from indic_transliteration.sanscript import transliterate
from gtts import gTTS
from IPython.display import Audio, display
import re
```

```python
def normalize_english_words(text):
    replacements = {
        "meeting": "मीटिंग",
        "office": "ऑफिस",
        "college": "कॉलेज",
        "call": "कॉल",
        "assignment": "असाइनमेंट",
        "project": "प्रोजेक्ट",
        "time": "टाइम"
    }

    words = text.split()
    result = []

    for w in words:
        clean = re.sub(r'[^\w]', '', w.lower())
        if clean in replacements:
            result.append(w.replace(clean, replacements[clean]))
        else:
            result.append(w)

    return " ".join(result)
```

```python
def roman_to_marathi(text):
    return transliterate(text, sanscript.ITRANS, sanscript.DEVANAGARI)
```

```python
def marathi_tts_autoplay(text):
    tts = gTTS(text=text, lang="mr")
    tts.save("output.mp3")
    display(Audio("output.mp3", autoplay=True))
```

```python
while True:
    print("\n👉 Enter Minglish text (press ENTER twice to finish):")

    lines = []
    while True:
        line = input()
        if line.strip() == "":
            break
        lines.append(line)

    user_input = " ".join(lines)

    if user_input.strip() == "":
        print("⚠️ No input given.")
    else:
        normalized = normalize_english_words(user_input)
        marathi_text = roman_to_marathi(normalized)



        print("\n📌 Marathi Text:")
        print(marathi_text)

        # 🔊 Speak
        marathi_tts_autoplay(marathi_text)

    # 🔁 Re-enter option
    choice = input("\n🔁 Do you want to enter text again? (yes/no): ").lower()
    if choice != "yes":
        print("👋 Exiting TTS system. धन्यवाद!")
        break
```
